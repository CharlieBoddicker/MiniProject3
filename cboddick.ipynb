```python
import pandas as pd

# Load both directions of validation
my_validation = pd.read_csv(
    "cboddick_validation_log.csv",
    sep=";",
    dtype=str,
    keep_default_na=False
)

phillip_validation = pd.read_csv(
    "pvickery_validation_log.csv",
    sep=";",
    dtype=str,
    keep_default_na=False
)

# Combine into 100 total comparisons
all_validation = pd.concat(
    [my_validation, phillip_validation],
    ignore_index=True
)

print("Total comparisons:", len(all_validation))

if "Field" in all_validation.columns:
    field_col = "Field"

elif "Field(one out of five listed above)" in all_validation.columns:
    field_col = "Field(one out of five listed above)"

else:
    raise KeyError("Could not find Field column.")



def course_kappa(percent_agree, num_labels):

    # Convert percent to proportion
    pa = percent_agree / 100

    # Chance agreement
    ps = 1 / num_labels

    # Cohen's Kappa
    kappa = (pa - ps) / (1 - ps)

    return round(kappa, 3)


# Number of possible labels for each categorical field
num_labels = {
    "HasRecovered": 2,       # Yes / No
    "ActivityPattern": 7,    # steady, rising, declining,
                             # U-shaped, stopped, cyclical, irregular
    "Currentstatus": 2       # Active / Inactive
}


fields = [
    "HasRecovered",
    "WhyRecovered",
    "ActivityPattern",
    "Currentstatus",
    "RecentThemes"
]

categorical_fields = [
    "HasRecovered",
    "ActivityPattern",
    "Currentstatus"
]


# ============================================================

notes = {

    "HasRecovered":
        "High agreement, The main disagreement involved whether a one-month gap was long enough to qualify as an inactivity gap.",

    "WhyRecovered":
        "Most differences came from how specifically the recovery evidence was described. Discussion helped combine or clarify interpretations.",

    "ActivityPattern":
        "This field caused the most disagreement. The main difficulty was distinguishing irregular patterns from declining, cyclical, steady, or stopped activity.",

    "Currentstatus":
        "Very high agreement. The main disagreement involved interpretation of the cutoff date used to classify a project as active or inactive.",

    "RecentThemes":
        "Most disagreements came from broad use of the Other category. Reviewing individual commit messages produced more specific theme classifications."
}



stats_rows = []

for field in fields:

    field_df = all_validation[
        all_validation[field_col] == field
    ].copy()

    total = len(field_df)


    pre_agree = (
        field_df[
            "PreDiscussAgreementstatus(Yes/No)"
        ]
        .str.strip()
        .str.lower()
        == "yes"
    )

    num_pre_agree = pre_agree.sum()

    num_disagreements = total - num_pre_agree

    percent_pre = round(
        (num_pre_agree / total) * 100,
        1
    )


    adjudicated = (
        field_df["AdjudicatedValue"]
        .fillna("")
        .str.strip()
    )

    resolved = (
        (~pre_agree)
        &
        (adjudicated != "")
    )

    num_resolved = resolved.sum()

    num_post_agree = (
        num_pre_agree
        +
        num_resolved
    )

    percent_post = round(
        (num_post_agree / total) * 100,
        1
    )



    if field in categorical_fields:

        kappa_pre = course_kappa(
            percent_pre,
            num_labels[field]
        )

        kappa_post = course_kappa(
            percent_post,
            num_labels[field]
        )

        author_post = (
            field_df["AuthorValue"]
            .copy()
        )

        validator_post = (
            field_df["ValidatorValue"]
            .copy()
        )

        for index in field_df.index:

            adj = str(
                field_df.loc[
                    index,
                    "AdjudicatedValue"
                ]
            ).strip()

            if adj != "":

                author_post.loc[index] = adj
                validator_post.loc[index] = adj

        kappa_post = cohen_kappa(
            author_post,
            validator_post
        )

        kappa_post = round(
            kappa_post,
            3
        )

    else:

        kappa_pre = "NA"
        kappa_post = "NA"


    stats_rows.append({

        "Field":
            field,

        "PercentAgree_Pre":
            percent_pre,

        "PercentAgree_Post":
            percent_post,

        "CohenKappa_Pre":
            kappa_pre,

        "CohenKappa_Post":
            kappa_post,

        "NumDisagreements":
            num_disagreements,

        "NumResolved":
            num_resolved,

        "Notes":
            notes[field]
    })



agreement_stats = pd.DataFrame(
    stats_rows
)



agreement_stats.to_csv(
    "cboddick_agreement_stats.csv",
    sep=";",
    index=False
)



print(
    "Created cboddick_agreement_stats.csv"
)

print(
    "Total comparisons:",
    len(all_validation)
)

print()

display(
    agreement_stats
)
```
